**Đề bài:** Theo thống kê các buổi trước, trung bình có 2 sinh viên nghỉ học mỗi buổi. Tính xác suất:
1. Không có sinh viên nào nghỉ học
2. Có nhiều nhất 2 sinh viên nghỉ học
3. Có ít nhất 1 sinh viên nghỉ học

## Mô hình

Gọi $X$ là số sinh viên nghỉ học trong một buổi. Đây là số lần xảy ra một sự kiện trong một khoảng cố định, với tần suất trung bình đã biết, nên ta mô hình hoá

$$X \sim \text{Poisson}(\lambda = 2), \qquad P(X = k) = \frac{e^{-\lambda}\,\lambda^k}{k!}, \quad k = 0, 1, 2, \dots$$

```python
import math
import numpy as np
from scipy.stats import poisson

lam = 2   # trung bình số sinh viên nghỉ mỗi buổi

# Hàm khối xác suất (PMF) của phân phối Poisson, viết theo công thức
def poisson_pmf(k, lam):
    return math.exp(-lam) * lam**k / math.factorial(k)

# Bảng PMF cho vài giá trị đầu
for k in range(6):
    print(f"P(X = {k}) = {poisson_pmf(k, lam):.4f}")
```

## 1. Không có sinh viên nào nghỉ học

$$P(X = 0) = e^{-2}$$

```python
p1 = poisson_pmf(0, lam)
print(f"P(X = 0) = {p1:.4f}")
print(f"Kiểm tra bằng scipy: {poisson.pmf(0, lam):.4f}")
```

## 2. Có nhiều nhất 2 sinh viên nghỉ học

$$P(X \le 2) = P(X=0) + P(X=1) + P(X=2) = e^{-2}\left(1 + 2 + \tfrac{2^2}{2!}\right) = 5e^{-2}$$

```python
p2 = sum(poisson_pmf(k, lam) for k in range(3))
print(f"P(X ≤ 2) = {p2:.4f}")
print(f"Kiểm tra bằng scipy: {poisson.cdf(2, lam):.4f}")
```

## 3. Có ít nhất 1 sinh viên nghỉ học

Dùng biến cố đối:

$$P(X \ge 1) = 1 - P(X = 0) = 1 - e^{-2}$$

```python
p3 = 1 - poisson_pmf(0, lam)
print(f"P(X ≥ 1) = {p3:.4f}")
print(f"Kiểm tra bằng scipy: {poisson.sf(0, lam):.4f}")
```

## Kiểm tra bằng mô phỏng Monte Carlo

Sinh ngẫu nhiên 1.000.000 buổi học theo Poisson(2) và so sánh tần suất thực nghiệm với kết quả lý thuyết.

```python
rng = np.random.default_rng(2026)
N = 1_000_000
samples = rng.poisson(lam, size=N)

print(f"Mô phỏng  P(X = 0) ≈ {np.mean(samples == 0):.4f}   (lý thuyết {p1:.4f})")
print(f"Mô phỏng  P(X ≤ 2) ≈ {np.mean(samples <= 2):.4f}   (lý thuyết {p2:.4f})")
print(f"Mô phỏng  P(X ≥ 1) ≈ {np.mean(samples >= 1):.4f}   (lý thuyết {p3:.4f})")
```
